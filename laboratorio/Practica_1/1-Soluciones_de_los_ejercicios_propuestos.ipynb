### Ejercicio 1

Definir una función `estima_prob_condicional_atributo` tal que:
- Reciba como argumentos:
  - Una matriz `atributo` con los valores, usando codificación _one-hot_, de un atributo discreto para todos los ejemplos de una determinada clase.
  - Una cantidad `k`, con valor 1 por defecto, de ejemplos virtuales adicionales a considerar para cada posible valor del atributo.
- Devuelva como resultado un vector con la estimación de máxima verosimilitud de la distribución de probabilidad del atributo condicionada a la clase.

### Solución del ejercicio 1

````python
def estima_prob_condicional_atributo(atributo, k=1):
    frecuencias = atributo.sum(axis=0)
    frecuencias = frecuencias + k
    print("frecuencias", frecuencias)
    return frecuencias / frecuencias.sum()
````

  - Un vector `objetivo` con los valores de un atributo objetivo discreto para todos los ejemplos.
  - Una cantidad `k`, con valor 1 por defecto, de ejemplos virtuales adicionales a considerar para cada posible valor del atributo.
- Devuelva como resultado una matriz que contenga las estimaciones de máxima verosimilitud de las distribuciones de probabilidad del atributo condicionadas a cada posible clase.

**Nota**: considérese el uso de la [función `unique`](https://numpy.org/doc/stable/reference/generated/numpy.unique.html) de Numpy para determinar los posibles valores del atributo objetivo.

### Solución del ejercicio 2

````python
def estima_prob_condicionales_atributo(atributo, objetivo, k=1):
    clases = np.unique(objetivo)
    print("clases", clases)
    matriz_de_frecuencias = np.empty((clases.shape[0], atributo.shape[1]))
    for i, clase in enumerate(clases):
        # objetivo == clase crea una máscara booleana que selecciona las filas que cumplen
        atributo_en_clase = atributo[objetivo == clase]
        matriz_de_frecuencias[i] = estima_prob_condicional_atributo(atributo_en_clase, k)
    return matriz_de_frecuencias
````

### Ejercicio 3

Definir una función `calcula_índice_de_Gini` tal que:
- Reciba como argumentos:
  - Un vector `objetivo` con los valores de un atributo objetivo discreto para un conjunto de ejemplos.
- Devuelva como resultado el índice de Gini, relativo al atributo objetivo, del conjunto de ejemplos.

### Solución del ejercicio 3

````python
tipo_de_consumo
````

````python
# formula 
def calcula_índice_de_Gini(objetivo):
    _, frecuencias = np.unique(objetivo, return_counts=True)
    print("frecuencias", frecuencias)
    frecuencias = frecuencias / frecuencias.sum()
    print("frecuencias normalizadas", frecuencias)
    return 1 - np.sum(np.square(frecuencias))  # O bien return 1 - np.sum(frecuencias ** 2)
````

### Ejercicio 4

Definir una función `calcula_impurezas_promedio` tal que:
- Reciba como argumentos:
  - Un vector `atributo` con los valores de un atributo numérico para un conjunto de ejemplos.
  - Un vector `objetivo` con los valores de un atributo objetivo discreto para el mismo conjunto de ejemplos.
- Devuelva como resultado un vector con las impurezas promedio de todas las particiones inducidas por los umbrales candidatos considerados en el algoritmo CART.

### Solución del ejercicio 4

````python
def calcula_impurezas_promedio(atributo, objetivo):
    # el atributo se emplea para encontrar los umbrales,
    # y con ellos encontrar las Y que son realmente las usadas en Gini
    total_ejemplos = atributo.shape[0]
    valores = np.unique(atributo)
    print("posibles valores", valores)
    umbrales = (valores[:-1] + valores[1:]) / 2
    impurezas_promedio = np.empty_like(umbrales)
    print("umbrales", umbrales)
    for i, umbral in enumerate(umbrales):
        print("\numbral", umbral)
        # selecciona las Y de los ejemplos que cumplen la condición de cada rama
        D_izq = objetivo[atributo <= umbral]
        print("izq", D_izq)
        # selecciona las Y de los ejemplos que cumplen la condición de cada rama
        D_der = objetivo[atributo > umbral]
        print("der", D_der)
        partición = (D_izq, D_der)
        índices_de_Gini = np.array([calcula_índice_de_Gini(D) for D in partición])
        print("índices de Gini", índices_de_Gini)
        tamaños_relativos = np.array([D.shape[0] for D in partición]) / total_ejemplos
        print("tamaños relativos", tamaños_relativos)
        impureza_promedio = np.average(índices_de_Gini, weights=tamaños_relativos)  # O bien np.sum(tamaños_relativos * índices_de_Gini)
        print("impureza promedio", impureza_promedio)
        impurezas_promedio[i] = impureza_promedio

    return impurezas_promedio
````

### Ejercicio 5

Definir una función `obtén_vecinos_más_cercanos` tal que:
- Reciba como argumentos:
  - Un vector `ejemplo` con los valores de todos los atributos, numéricos, para un determinado ejemplo.
  - Una matriz `conjunto_de_ejemplos` con los valores de todos los atributos, numéricos, para cada ejemplo de un determinado conjunto.
  - Una cantidad `k` de vecinos a obtener.
- Devuelva como resultado una matriz con los `k` ejemplos de `conjunto_de_ejemplos` más cercanos a `ejemplo`, usando la distancia euclídea como medida de cercanía.

**Nota**: considérese el uso de la [función `argsort`](https://numpy.org/doc/stable/reference/generated/numpy.argsort.html) de Numpy para obtener los índices de los vecinos más cercanos.

### Solución del ejercicio 5

````python
def obtén_vecinos_más_cercanos(ejemplo, conjunto_de_ejemplos, k):
    # Como la raíz cuadrada es una función creciente, para determinar los vecinos más cercanos basta
    # calcular las distancias euclídeas cuadráticas
    raices = np.square(conjunto_de_ejemplos - ejemplo)
    print("raices\n", raices)
    # axis=1 computa la suma de cada fila
    # se devuelve un array con el size de las filas
    distancias_euclídeas_cuadráticas = np.sum(raices, axis=1)
    print("distancias\n", distancias_euclídeas_cuadráticas)
    # busca los índices de las filas ordenados de menor a mayor distancia
    índices_vecinos_ordenados = np.argsort(distancias_euclídeas_cuadráticas)
    print("índices vecinos ordenados\n", índices_vecinos_ordenados)
    índices_vecinos_más_cercanos = índices_vecinos_ordenados[:k]
    return conjunto_de_ejemplos[índices_vecinos_más_cercanos]
````

### Ejercicio 6

Definir una función `codifica_atributo_discreto` tal que:
- Reciba como argumentos:
  - Una _Series_ de Pandas `atributo` con los valores de un atributo discreto para cada ejemplo de un determinado conjunto.
- Devuelva como resultado una **nueva** _Series_ de Pandas con esos valores codificados (en orden) numéricamente a partir de 0.

**Nota**: considérese el uso del [método `sort`](https://numpy.org/doc/stable/reference/generated/numpy.ndarray.sort.html) de los _arrays_ de Numpy y de la [función `select`](https://numpy.org/doc/stable/reference/generated/numpy.select.html) de Numpy.

### Solución del ejercicio 6

````python
def codifica_atributo_discreto(atributo):
    valores = atributo.unique()
    valores.sort()
    print("valores puros\n", [atributo == valor for valor in valores])
    
    valores_codificados = np.select(
        # A los indices que salgan a True
        [atributo == valor for valor in valores],
        # Se le asocia el valor que se pase en range()
        range(len(valores))
    )
    
    print("valores_codificados\n", valores_codificados)

    # se sigue usando el mismo index: atributo.index
    
    return pd.Series(valores_codificados, index=atributo.index)

# Otra posible forma de resolver el ejercicio

# def codifica_atributo_discreto(atributo):
#     atributo_codificado = pd.Series(0, index=atributo.index, dtype='int64')
#     valores = atributo.unique()
#     valores.sort()
#     for i, valor in enumerate(valores):
#         atributo_codificado[atributo == valor] = i
#     return atributo_codificado
````

### Ejercicio 7

Definir una función `normaliza_atributo_numérico` tal que:
- Reciba como argumentos:
  - Una _Series_ de Pandas `atributo` con los valores de un atributo numérico para cada ejemplo de un determinado conjunto.
- Devuelva como resultado una **nueva** _Series_ de Pandas con esos valores normalizados al intervalo [0, 1].

### Solución del ejercicio 7

````python
def normaliza_atributo_numérico(atributo):
    m = atributo.min()
    M = atributo.max()
    return (atributo - m) / (M - m)
````

#### Ejemplos del funcionamiento pretendido
